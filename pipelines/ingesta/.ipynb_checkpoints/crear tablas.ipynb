```python
# ================================================================================
# PASO 1: IMPORTACIÓN DE LIBRERÍAS
# ================================================================================
import os
import psycopg2
from psycopg2.extras import execute_values
from dotenv import load_dotenv
from pathlib import Path
import pandas as pd
```

```python

# ================================================================================
# PASO 2: CARGA DE VARIABLES Y CONFIGURACIÓN
# ================================================================================
load_dotenv()

db_host = os.getenv("HOST_POSTGRES_NBK") 
db_port = os.getenv("PORT_POSTGRES")
db_name = os.getenv("POSTGRES_DB")
db_user = os.getenv("POSTGRES_USER")
db_password = os.getenv("POSTGRES_PASSWORD")
print("Base de datos conectada:", db_name)

# Crear la conexión
conn = psycopg2.connect(
    host=db_host,
    port=db_port,
    dbname=db_name,
    user=db_user,
    password=db_password
)
cur = conn.cursor()
```

```python

# ================================================================================
# PASO 3: CREA LA TABLAS EN LA BASE DE DATOS
# ================================================================================
### crea la tabla de dim_obra_social
cur.execute("""
-- dimensiones
CREATE TABLE IF NOT EXISTS dim_obra_social (
    obra_social_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    nombre TEXT UNIQUE,
    tipo TEXT,
    cobertura TEXT
);
""")

### Crear tabla dim_fecha si no existe
cur.execute("""
CREATE TABLE IF NOT EXISTS dim_fecha (
    fecha_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    fecha DATE UNIQUE,
    año INT,
    mes INT,
    dia INT,
    nombre_dia TEXT,
    es_fin_de_semana BOOLEAN,
    feriado BOOLEAN
);
""")
### crea la tabla de dim_paciente
cur.execute("""
CREATE TABLE IF NOT EXISTS dim_paciente (
    paciente_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    id INT UNIQUE,
    nombre TEXT,
    edad INTEGER
);
""")
### crea la tabla de dim_turno
cur.execute("""
CREATE TABLE IF NOT EXISTS dim_turno (
    turno_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    especialidad TEXT NOT NULL,
    duracion_min INT NOT NULL,
    UNIQUE (especialidad, duracion_min)
);
""")

### crea la tabla de fac_turno_paciente
cur.execute("""
CREATE TABLE IF NOT EXISTS fact_turno_paciente (
    fact_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,                     -- Clave primaria de la tabla, autogenerada (tipo surrogate key).
                                                                                 -- Se usa para identificar de forma única cada registro de la tabla de hechos.
    paciente_sk BIGINT NOT NULL REFERENCES dim_paciente(paciente_sk),            -- Clave foránea que apunta a la dimensión de pacientes.
                                                                                 -- Relaciona el hecho con un paciente específico.
    obra_social_sk BIGINT NOT NULL REFERENCES dim_obra_social(obra_social_sk),   -- Clave foránea que apunta a la dimensión de obra social.
    fecha_sk BIGINT NOT NULL REFERENCES dim_fecha(fecha_sk),   
                                                                                 -- Clave foránea que apunta a la dimensión de fechas.
    turno_sk BIGINT NOT NULL REFERENCES dim_turno(turno_sk),                     -- Clave foránea que apunta a la dimensión de turno.
                                                                                 -- Puede contener detalles como horario, duración, etc.
    edad INT,
    creado_en TIMESTAMPTZ DEFAULT NOW(),
    UNIQUE(paciente_sk, obra_social_sk, fecha_sk, turno_sk)                      -- Marca temporal automática de cuándo se insertó el registro en la tabla.
);
""")
conn.commit()
```

```python

# ================================================================================
# PASO 4: # CONEXION A LA BASE DE DATOS Y A LOS CSV
# ================================================================================

conn = psycopg2.connect(
    host=db_host,
    port=db_port,
    dbname=db_name,
    user=db_user,
    password=db_password
)
cur = conn.cursor()

# Ruta base de los CSV
ruta_base = Path("/home/jovyan/datos/csv")


# Cargar cada CSV en su DataFrame
df_obra_social = pd.read_csv(ruta_base / "dim_obra_social.csv")
df_fecha = pd.read_csv(ruta_base / "dim_fecha.csv")
df_paciente = pd.read_csv(ruta_base / "dim_paciente.csv")
df_turno = pd.read_csv(ruta_base / "dim_turno.csv")
```

```python

# ================================================================================
# PASO 5: # PREPARA LOS DATOS LIMPIOS,CONSISTENTES PARA CARGARLOS EN LA BASE DE DATOS DIM_FECHA
# ================================================================================

df_fecha['fecha'] = pd.to_datetime(df_fecha['fecha']).dt.date
df_fecha['es_fin_de_semana'] = df_fecha['es_fin_de_semana'].astype(bool)

feriados_arg = pd.to_datetime([
    '2023-01-01',  # Año Nuevo
    '2023-02-20',  # Carnaval
    '2023-02-21',  # Carnaval
    '2023-03-24',  # Día de la Memoria
    '2023-04-02',  # Día del Veterano (trasladado)
    '2023-04-07',  # Viernes Santo
    '2023-05-01',  # Día del Trabajador
    '2023-05-25',  # Revolución de Mayo
    '2023-06-17',  # Paso a la Inmortalidad Gral. Güemes
    '2023-06-20',  # Paso a la Inmortalidad Gral. Belgrano
    '2023-07-09',  # Independencia
    '2023-08-21',  # Paso San Martín (trasladado)
    '2023-10-16',  # Diversidad cultural (trasladado)
    '2023-11-20',  # Soberanía nacional
    '2023-12-08',  # Inmaculada Concepción
    '2023-12-25'   # Navidad
]).date

# Aplicar feriados
df_fecha['feriado'] = df_fecha['fecha'].isin(feriados_arg)

# quedarnos solo con columnas relevantes y quitar duplicados por fecha
df = df_fecha[['fecha', 'nombre_dia', 'es_fin_de_semana', 'feriado']].drop_duplicates(subset=['fecha'])

# Crear tuplas para insertar
records = [
    (
        fecha,
        nombre_dia.strip() if isinstance(nombre_dia, str) else None,
        bool(es_fin),
        bool(feriado)
    )
    for fecha, nombre_dia, es_fin, feriado in df.itertuples(index=False, name=None)
]
# ------------------------------------------------------------
# Sentencia SQL para insertar o actualizar datos en la tabla dim_fecha
# Si la fecha ya existe (conflicto), se actualizan los campos correspondientes
# ------------------------------------------------------------
query = """
INSERT INTO dim_fecha (fecha, nombre_dia, es_fin_de_semana, feriado)
VALUES %s
ON CONFLICT (fecha) DO UPDATE
SET 
    nombre_dia = EXCLUDED.nombre_dia,
    es_fin_de_semana = EXCLUDED.es_fin_de_semana,
    feriado = EXCLUDED.feriado;
"""
execute_values(cur, query, records)
conn.commit()

cur.close()
conn.close()
```

```python

# ================================================================================
# PASO 6: # PREPARA LOS DATOS LIMPIOS,CONSISTENTES PARA CARGARLOS EN LA BASE DE DATOS DIM_PACIENTE
# ================================================================================
# Crear la conexión
cur = conn.cursor()

df_paciente = pd.read_csv(ruta_base / "dim_paciente.csv")

df = df_paciente.rename(columns=str.lower)

df_upsert = (
    df[['id', 'nombre', 'edad']]
    .drop_duplicates(subset=['id'], keep='last')
    .astype({'id': int, 'edad': int, 'nombre': str})
)

records = list(df_upsert.itertuples(index=False, name=None))  # tuplas (id, nombre, edad)

if not records:
    print("No hay registros para procesar.")
else:
    query = """
    INSERT INTO public.dim_paciente (id, nombre, edad)
    VALUES %s
    ON CONFLICT (id) DO UPDATE
    SET nombre = EXCLUDED.nombre,
        edad = EXCLUDED.edad;
    """
    with conn.cursor() as cur:
        execute_values(cur, query, records, page_size=100)
    conn.commit()
    print(f"Upsert completado: {len(records)} registros procesados.")
```

```python
# %%
# ================================================================================
# PASO 7: # PREPARA LOS DATOS LIMPIOS,CONSISTENTES PARA CARGARLOS EN LA BASE DE DATOS DIM_OBRA_SOCIAL
# ================================================================================
# Crear la conexión
cur = conn.cursor()

# 2) Normalizamos nombres de columna a minúsculas
df_obra_social = df_obra_social.rename(columns=str.lower)

df_upsert = (
    df_obra_social[['nombre', 'tipo', 'cobertura']]
    .drop_duplicates(subset=['nombre'], keep='last')
    .astype({'nombre': str, 'tipo': str, 'cobertura': str})
)

# 4) Armamos la lista de tuplas
records = list(df_upsert.itertuples(index=False, name=None))
# -> [( 'OS A', 'pública', 'total' ), ( 'OS B', 'privada', 'parcial' ), …]

if not records:
    print("No hay registros para insertar/actualizar en dim_obra_social.")
else:
    # 5) Query de upsert
    query = """
    INSERT INTO dim_obra_social (nombre, tipo, cobertura)
    VALUES %s
    ON CONFLICT (nombre) DO UPDATE
      SET tipo     = EXCLUDED.tipo,
          cobertura = EXCLUDED.cobertura;
    """
    # 6) Ejecutamos
    with conn.cursor() as cur:
        execute_values(cur, query, records, page_size=100)
    conn.commit()
    print(f"Upsert completado: {len(records)} registros procesados en dim_obra_social.")
```

```python
# %%
# ================================================================================
# PASO 8: # PREPARA LOS DATOS LIMPIOS,CONSISTENTES PARA CARGARLOS EN LA BASE DE DATOS DIM_TURNO
# ================================================================================
# Crear la conexión
cur = conn.cursor()

#  Seleccionar sólo las columnas necesarias y quitar duplicados
df_turno = (
    df_turno[['especialidad', 'duracion_min']]
    .drop_duplicates(subset=['especialidad', 'duracion_min'], keep='last')
    .astype({'especialidad': str, 'duracion_min': int})
)

#  Armado de tuplas para execute_values
records = list(df_turno.itertuples(index=False, name=None))


if not records:
    print("No hay registros para procesar en dim_turno.")
else:
    #  Query de inserción con manejo de duplicados
    query = """
    INSERT INTO dim_turno (especialidad, duracion_min)
    VALUES %s
    ON CONFLICT (especialidad, duracion_min) DO NOTHING;
    """
    with conn.cursor() as cur:
        execute_values(cur, query, records, page_size=100)
    conn.commit()
    print(f"Cargados/upserted {len(records)} registros en dim_turno.")
```
