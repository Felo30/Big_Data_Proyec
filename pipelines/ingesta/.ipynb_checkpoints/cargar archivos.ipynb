```python
# ================================================================================
# PASO 1: IMPORTACIÓN DE LIBRERÍAS
# ================================================================================
import os
import psycopg2
from dotenv import load_dotenv
from pathlib import Path
import pandas as pd
```

```python
## ruta del archivo
ruta_csv = Path("/home/jovyan/datos/csv/pacientesBronce_csv.csv")
```

```python
### cargar el archivo csv 
df = pd.read_csv(ruta_csv)
```

```python

# ================================================================================
# PASO 2: CARGA DE VARIABLES Y CONFIGURACIÓN
# ================================================================================
load_dotenv()

# Extraer variables de conexión para PostgreSQL

### Busca la variable de entorno HOST_POSTGRES_NBK. Si existe y no está vacía, la usa.##

db_host = os.getenv("HOST_POSTGRES_NBK") or os.getenv("HOST_POSTGRES")   
db_port = os.getenv("PORT_POSTGRES")
db_name = os.getenv("POSTGRES_DB")
db_user = os.getenv("POSTGRES_USER")
db_password = os.getenv("POSTGRES_PASSWORD")
print("Base de datos conectada:", db_name)

# Crear la conexión
conn = psycopg2.connect(
    host=db_host,
    port=db_port,
    dbname=db_name,
    user=db_user,
    password=db_password
)
```

```python
# ================================================================================
# PASO 3: CREA LA TABLAS EN LA BASE DE DATOS
# ================================================================================
# Crear un cursor para ejecutar consultas
cur = conn.cursor()

# Crear tabla pacientes
create_table_query = """
CREATE TABLE IF NOT EXISTS pacientesBronce (
    id INT PRIMARY KEY,
    nombre VARCHAR(100),
    edad INT,
    obra_social VARCHAR(100),
    fecha_turno TIMESTAMP
);
"""

cur.execute(create_table_query)
conn.commit()

print("Tabla pacientesBronce creada correctamente")

cur.close()
conn.close()
```

```python
# ================================================================================
# PASO 4: # CONEXION A LA BASE DE DATOS Y AL CSV
# ================================================================================

# 1. Convertir fecha y limpiar NaNs
df['fecha_turno'] = pd.to_datetime(df['fecha_turno'], errors='coerce')
df['obra_social'] = df['obra_social'].where(pd.notna(df['obra_social']), None)

# 2. Tomar solo los primeros 100 registros
df_limited = df.head(100)

# 3. Preparar lista de tuplas, convirtiendo Timestamp a datetime.datetime
data = []
for _, row in df_limited.iterrows():
    fecha = (row['fecha_turno'].to_pydatetime() 
             if pd.notna(row['fecha_turno']) else None)
    edad  = int(row['edad']) if pd.notna(row['edad']) else None
    data.append((
        int(row['id']),
        row['nombre'],
        edad,
        row['obra_social'],
        fecha
    ))

# 4. Conectar a Postgres e insertar
conn = psycopg2.connect(
    host=db_host,
    port=db_port,
    dbname=db_name,
    user=db_user,
    password=db_password
)
cur = conn.cursor()

# ================================================================================
# PASO 5: # INSERTA LOS DATOS EN LA TABLA DE PACIENTESBRONCE
# ================================================================================

cur.executemany("""
    INSERT INTO pacientesBronce (id, nombre, edad, obra_social, fecha_turno)
    VALUES (%s, %s, %s, %s, %s)
    ON CONFLICT (id) DO NOTHING;
""", data)
conn.commit()

# 5. Verificar en base de datos que tienes 20 filas
cur.execute("SELECT * FROM pacientesBronce ORDER BY id LIMIT 100;")
rows = cur.fetchall()
print(f"Se recuperaron {len(rows)} filas de la tabla pacientes:")
for r in rows:
    print(r)

cur.close()
conn.close()
```
